# Targeted Employment Support System

## Overview

This project implements a semantic search and recommendation system designed to match job seekers with relevant resources. The system uses resumes and natural language queries as input and leverages machine learning and natural language processing (NLP) techniques to generate personalized resource recommendations. The system is powered by embeddings generated with OpenAI’s API and incorporates various libraries and tools for data preprocessing, embedding storage, and similarity matching.

## Key Functionalities

1. **Input Handling:**
   - Accepts resumes or natural language queries from users.
   - Processes resumes using NLP techniques to extract key information (e.g., skills, industries, barriers) and generates concise, structured queries.

2. **Resource Description Enhancement:**
   - Enriches resource descriptions using GPT-based content enhancement to add context (e.g., barriers addressed, geographic scope).

3. **Embedding Generation:**
   - Converts user queries and resource descriptions into high-dimensional embeddings using OpenAI's text-embedding-ada-002 model.

4. **Similarity Matching:**
   - Performs cosine similarity calculations between user input embeddings and stored resource embeddings to identify the most relevant matches.

5. **Result Display:**
   - Presents recommendations to users through an intuitive Streamlit interface, showing detailed resource descriptions tailored to their needs.

## Endpoints and Libraries Used

### Endpoints:

1. **OpenAI API:**
   - Used for embedding generation (`text-embedding-ada-002`).
   - Key functionality: Converts text data into 1,536-dimensional embeddings for semantic matching.
   - Endpoint: `openai.Embedding.create`.

2. **ChromaDB:**
   - Vector database used to store and retrieve embeddings efficiently.
   - Key functionality: Enables nearest-neighbor search to retrieve relevant resources based on query embeddings.

### Libraries and Tools:

- **Streamlit**: Framework for creating the user interface (UI) to upload resumes and display recommendations.
- **PyPDF2**: Used for extracting text from uploaded PDF resumes.
- **Pandas**: For data preprocessing and managing datasets of resources and candidates.
- **OpenAI API Library**: Interfaced with OpenAI’s embedding endpoint.
- **ChromaDB**: Facilitated the storage and retrieval of embeddings.
- **Hashlib**: Used to generate unique resource IDs for tracking purposes.
- **Numpy**: Assisted in numerical operations, including cosine similarity calculations.

## How It Works

### Data Preprocessing:

- **Resumes:**
  - Extract text using PyPDF2 or process plain text files.
  - Summarize resumes into concise queries using GPT-based summarization.

- **Resources:**
  - Enhance descriptions by appending attributes such as industry relevance, barriers addressed, and geographic applicability.

### Embedding Workflow:
1. Generate embeddings for resumes/queries and resource descriptions using OpenAI's API.
2. Store resource embeddings in ChromaDB for fast similarity searches.

### Matching:
- Calculate cosine similarity between query embeddings and resource embeddings.
- Rank resources by relevance and present top matches to the user.

## Code Structure

### File 1: `app.py`
- **Purpose**: Implements the Streamlit-based user interface.
- **Key Functions**:
  - `extract_concise_query`: Summarizes resume text into a concise natural language query.
  - `enhance_description`: Enhances resource descriptions using GPT-based content generation.
  - `generate_embedding`: Calls OpenAI API to generate embeddings for text data.
  - `initialize_collection`: Prepares the ChromaDB collection for resource embeddings.

### File 2: `risekit.py`
- **Purpose**: Implements the backend functionalities for embedding generation, storage, and similarity matching.
- **Key Functions**:
  - `generate_resource_id`: Generates unique IDs for resources using hashlib.
  - `query_embeddings`: Queries ChromaDB to retrieve relevant resources based on cosine similarity.

## How to Run the System

1. **Set Up the Environment:**
   - Install required libraries:
     ```bash
     pip install streamlit openai numpy pandas chromadb
     ```
   - Set the OpenAI API key as an environment variable:
     ```bash
     export OPENAI_API_KEY="your_openai_api_key"
     ```

2. **Launch the Streamlit App:**
   - Run the app.py file:
     ```bash
     streamlit run app.py
     ```

3. **Input Options:**
   - Upload a resume in `.pdf` or `.txt` format, or enter a query directly.

4. **View Recommendations:**
   - The app will display the top 5 recommended resources based on semantic similarity.
